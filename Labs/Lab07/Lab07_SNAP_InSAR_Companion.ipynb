## 3. Check the scale of a fringe · optional

One complete color cycle on a full wrapped phase scale is $2\pi$ radians. The radar signal travels to the ground and back, so the equivalent change in line of sight distance is half a wavelength if the phase change comes entirely from motion. Run the cell to see that scale for Sentinel-1. [Copernicus Sentinel-1 instrument description](https://sentiwiki.copernicus.eu/web/s1-mission)

```python
from math import pi

wavelength_cm = 5.55
phase_change_rad = 2 * pi  # One full fringe cycle
los_change_cm = wavelength_cm * abs(phase_change_rad) / (4 * pi)

print(f"If caused by motion, one full phase cycle represents about {los_change_cm:.1f} cm along the line of sight.")
```
